```python
# @title ⚙️ Step 3: Hyper-Speed Cinema Engine (NVENC Hardware Acceleration)
import re, time, random, urllib.request, subprocess
from google import genai
from google.genai import types
from groq import Groq

# --- 1. TOPIC BRAIN ---
def discover_topic(channel_id: str) -> str:
    ch_meta = CONFIG["CHANNELS"][channel_id]
    client = Groq(api_key=CONFIG["GROQ_API_KEYS"][0])
    prompt = (
        f"Generate 1 brand-new, soul-shaking YouTube topic in pure Hindi Devanagari for {ch_meta['name']}. "
        f"Focus on: {ch_meta['niche']}. ZERO modern English words. Return ONLY the title text in Hindi."
    )
    res = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[{"role": "user", "content": prompt}],
        temperature=0.8
    )
    return res.choices[0].message.content.strip().replace('"', '')

# --- 2. SCRIPT ARCHITECT (10 Chapters, 4,000+ Words) ---
def generate_script(topic: str) -> list:
    client = Groq(api_key=CONFIG["GROQ_API_KEYS"][0])
    prompt = (
        f"Write a 10-chapter long-form masterpiece in 100% PURE LITERARY HINDI DEVANAGARI for video topic: '{topic}'.\n"
        f"RULES: Minimum 350 words per chapter. ZERO English words, ZERO Sanskrit recital in voice.\n"
        f"Return ONLY a valid JSON array of objects: [{{'chapter_num': 1, 'title': 'शीर्षक', 'exposition': 'गहरा विवेचन...', 'visual_queries': ['meditation', 'ancient temple']}}]"
    )
    res = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[{"role": "user", "content": prompt}],
        temperature=0.7,
        response_format={"type": "json_object"}
    )
    raw = res.choices[0].message.content
    parsed = json.loads(raw)
    chapters = parsed.get("chapters") or parsed.get("data") or list(parsed.values())[0]
    return chapters if isinstance(chapters, list) else []

# --- 3. GOOGLE GEMINI REALISTIC TTS ('Charon') + Shure SM7B DSP ---
def synthesize_gemini_voice(text: str, out_path: str) -> float:
    key = random.choice(CONFIG["GEMINI_API_KEYS"])
    client = genai.Client(api_key=key)
    raw_pcm = out_path + ".pcm"
    cfg = types.GenerateContentConfig(
        response_modalities=["AUDIO"],
        speech_config=types.SpeechConfig(
            voice_config=types.VoiceConfig(
                prebuilt_voice_config=types.PrebuiltVoiceConfig(voice_name=CONFIG["VOICE_NAME"])
            )
        )
    )
    resp = client.models.generate_content(model="gemini-2.5-flash-preview-tts", contents=text, config=cfg)
    pcm_data = b""
    for part in resp.candidates[0].content.parts:
        if part.inline_data:
            pcm_data += part.inline_data.data
    with open(raw_pcm, "wb") as f:
        f.write(pcm_data)
    
    dsp = "aresample=48000,highpass=f=75,equalizer=f=180:t=q:w=1.2:g=3.5,equalizer=f=450:t=q:w=1.5:g=-1.8,equalizer=f=3200:t=q:w=1.1:g=2.5,equalizer=f=6800:t=q:w=2.0:g=-3.0,lowpass=f=13500,compand=attacks=0.015:decays=0.12:points=-80/-80|-45/-32|-24/-17|-12/-10|0/-7:gain=2.5,loudnorm=I=-14:TP=-1.0:LRA=9"
    cmd = ["ffmpeg", "-y", "-f", "s16le", "-ar", "24000", "-ac", "1", "-i", raw_pcm, "-af", dsp, out_path]
    subprocess.run(cmd, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL, check=True)
    
    probe = subprocess.run(["ffprobe", "-v", "error", "-show_entries", "format=duration", "-of", "default=noprint_wrappers=1:nokey=1", out_path], capture_output=True, text=True)
    return float(probe.stdout.strip())

# --- 4. 24K GOLD SUBTITLES GENERATOR ---
def create_subtitles(text: str, duration: float, srt_path: str):
    words = text.split()
    chunks = [" ".join(words[i:i+5]) for i in range(0, len(words), 5)]
    t_per = duration / max(len(chunks), 1)
    entries = []
    cur = 0.5
    for idx, c in enumerate(chunks, 1):
        start_s = f"{int(cur//3600):02d}:{int((cur%3600)//60):02d}:{int(cur%60):02d},{int((cur%1)*1000):03d}"
        end_s = f"{int((cur+t_per)//3600):02d}:{int(((cur+t_per)%3600)//60):02d}:{int((cur+t_per)%60):02d},{int(((cur+t_per)%1)*1000):03d}"
        entries.append(f"{idx}\n{start_s} --> {end_s}\n{c}\n")
        cur += t_per
    with open(srt_path, "w", encoding="utf-8") as f:
        f.write("\n".join(entries))

# --- 5. T4 GPU HARDWARE ACCELERATED RENDER (NVENC) ---
def render_chapter_nvenc(raw_video: str, audio_path: str, srt_path: str, out_path: str, duration: float):
    sub_filter = f"subtitles={srt_path}:force_style='{CONFIG["SUBTITLE_STYLE"]}'"
    cmd = [
        "ffmpeg", "-y",
        "-i", raw_video,
        "-i", audio_path,
        "-filter_complex", f"[0:v]{sub_filter}[v_out];[1:a]volume=1.50[a_out]",
        "-map", "[v_out]", "-map", "[a_out]",
        "-t", f"{duration:.2f}",
        "-c:v", "h264_nvenc", "-preset", "p4", "-cq", "22",
        "-c:a", "aac", "-b:a", "192k",
        out_path
    ]
    subprocess.run(cmd, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL, check=True)

print("✅ HyperEngine Core Initialized with Tesla T4 NVENC GPU Acceleration!")
```
